### Manual binary conversion

```python
def to_binary(n):
    bits = []
    while n > 0:
      bits.append(n % 2)
      n //= 2
    return bits[::-1]
```

```python
to_binary(10)
```

### Triangle

```python
n = 4
for i in range(1, n + 1):
    for j in range(i):
      print('*', end = ' ')
    print()
```

## BinaryGap
### Find longest sequence of zeros in binary representation of an integer.

## Binary Gap

A **binary gap** is the longest sequence of consecutive zeros between two `1`s
in the binary representation of a positive integer N.

For example:
- `N = 9` → binary `1001` → gap of length **2**
- `N = 529` → binary `1000010001` → gaps of length **4** and **3**
- `N = 32` → binary `100000` → **no gap** (no closing `1`)

### Task
Given a positive integer N, return the length of its longest binary gap.
Return `0` if no binary gap exists.

**Example:** `N = 1041` → binary `10000010001` → answer is `5`

### Constraints
- `1 <= N <= 2,147,483,647`

```python
def solution(N):
  binary = bin(N)[2:]         # Covert a number to binary
  current_gap = 0             # initial current_gap
  max_gap = 0                 # initial max_gap
  found_one = False           # inital found_one as reset
  for bit in binary:
    if bit == '1':
      found_one = True        # reset count when bit = 1
      max_gap = max(max_gap, current_gap)   # compare max between max_gap and current_gap
      current_gap = 0                       # when bit = 1 , don't count current_gap
    elif bit == '0':
      if found_one:                         # when bit = 1, if found_one = true
        current_gap += 1                    # count current_gap when bit = 0
  return max_gap                            # return max_gap
```

```python
print(solution(9))
print(solution(529))
print(solution(20))
print(solution(15))
print(solution(32))

```

```python
print(solution(1))
print(solution(2_147_483_647))
print(solution(100000000 ))
```
