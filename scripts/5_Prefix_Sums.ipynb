# Mushroom Picker - Prefix Sums

A picker at position k must make exactly m moves to maximize mushrooms collected.

Insight: The optimal path never reverses direction more than once.

```python
def prefix_sum(A):
    """Build prefix sum array in O(n). P[0] = 0 by covention."""
    n = len(A)
    P = [0] * (n + 1)
    for k in range(1, n + 1):
        P[k] = P[k - 1] + A[k - 1]
    return P

def count_total(P: List[int], x: int, y: int) -> int:
    """Sum of slice [x..y] in O(1) using prefix sums."""
    return P[y +  1] - P[x]

def mushroom_picker(A: List[int], k: int, m: int) -> int:
    """
    Returns the maximum mushrooms collectable starting at index k with exactly m moves.
    Startegy: go p steps one direction, then (m - p) steps the other.
    Since direction only reverses once, we just enumerate all p and take max.
    Time: O(n + m)
    """
    n = len(A)
    pref = prefix_sum(A)
    result = 0

    # Case 1: move left first, then right
    for p in range(min(m, k) + 1):
        left_pos = k - p
        right_pos = min(n - 1, max(k, k + m - 2 * p))
        result = max(result, count_total(pref, left_pos, right_pos))
    
    # Case 2: move right first, then left
    for p in range(min(m + 1, n - k)):
        right_pos = k + p
        left_pos = max(0, min(k, k - (m - 2 * p)))
        result = max(result, count_total(pref, left_pos, right_pos))

    return result

if __name__ == "__main__":
    A = [2, 3, 7, 5, 1, 3, 9]
    k = 4
    m = 6 

    print(f"Array:      {A}")
    print(f"Start pos:  k = {k}")
    print(f"Moves:      m = {m}")
    print(f"Max mushrooms: {mushroom_picker(A, k, m)}")
    # Expected: 25  (path: 4->3->2->3->4->5->6 -> 1+5+7+3+9 = 25)
```

# PassingCars

Count the number of passing cars on the road.

## Problem Statement

A single-lane communication channel processes a stream of $N$ data packets moving in opposite directions.

You are given a non-empty array $A$ of $N$ integers where:
* **`0`** represents a packet moving **East**.
* **`1`** represents a packet moving **West**.

A **"signal collision"** occurs between any pair of packets $(P, Q)$ such that $0 \le P < Q < N$, where packet $P$ is moving **Right** (`0`) and packet $Q$ is moving **Left** (`1`).

Write a function `solution(A)` that returns the total count of signal collisions. If the number of collisions exceeds $1,000,000,000$, return `-1`.

### Example

* **Input:** $A = [0, 1, 0, 1, 1]$
* **Output:** `5`
* **Explanation:** The pairs of indices causing collisions are $(0, 1)$, $(0, 3)$, $(0, 4)$, $(2, 3)$, and $(2, 4)$.

### Constraints

* $N$ is an integer within the range $[1..100,000]$.
* Each element of array $A$ is either `0` or `1`.


```python
def solution(A):
    east_cars = 0
    passing_cars = 0
    LIMIT = 1_000_000_000

    for car in A:
        if car == 0: 
            east_cars += 1
        else:
            passing_cars += east_cars
            if passing_cars > LIMIT:
                return -1
    return passing_cars
            
```

```python
import time
import unittest


class TestPassingCars(unittest.TestCase):

    def check_case(self, A, expected):
        start = time.perf_counter()
        result = solution(A)
        runtime = time.perf_counter() - start

        a_preview = (
            str(A)
            if len(A) <= 10
            else f"{A[:10]} ... [N={len(A)}]"
        )

        print()
        print("=" * 70)
        print(f"Input      : {a_preview}")
        print(f"Output     : {result}")
        print(f"Expected   : {expected}")
        print(f"Runtime    : {runtime:.8f}s")
        print(f"Complexity : O(N)")
        print(f"Status     : {'PASS' if result == expected else 'FAIL'}")
        print("=" * 70)

        self.assertEqual(result, expected)

    def test_example(self):
        """Standard example from the problem."""
        self.check_case([0, 1, 0, 1, 1], 5)

    def test_no_passing_cars(self):
        """All cars move East."""
        self.check_case([0, 0, 0], 0)

    def test_all_west(self):
        """All cars move West."""
        self.check_case([1, 1, 1], 0)

    def test_one_passing_pair(self):
        """One East car followed by one West car."""
        self.check_case([0, 1], 1)

    def test_multiple_east_cars(self):
        """Multiple East cars followed by one West car."""
        self.check_case([0, 0, 1], 2)

    def test_multiple_west_cars(self):
        """One East car followed by multiple West cars."""
        self.check_case([0, 1, 1], 2)

    def test_single_car(self):
        """Smallest possible input."""
        self.check_case([0], 0)

    def test_exceeds_limit(self):
        """Number of passing cars exceeds 1,000,000,000."""
        A = [0] * 50_001 + [1] * 50_000
        self.check_case(A, -1)


if __name__ == "__main__":
    suite = unittest.TestLoader().loadTestsFromTestCase(TestPassingCars)
    unittest.TextTestRunner(verbosity=2).run(suite)
```

# CountDiv

## Problem Statement
A data streaming pipeline processes batch jobs identified by sequential integer IDs starting from $A$ to $B$ (inclusive). To monitor pipeline performance, the system periodically collects health telemetry for every job ID that is an exact multiple of a heartbeat frequency $K$.

Write a function `solution(A, B, K)` that returns the total number of job IDs within the range $[A..B]$ that trigger health telemetry collection, i.e.:

$$\{ i : A \le i \le B, i \bmod K = 0 \}$$

## Example
Input: $A = 6$, $B = 11$, $K = 2$  
Output: `3`  
Explanation: The job IDs in the range $[6..11]$ divisible by $2$ are $6$, $8$, and $10$.

## Constraints
- $A$ and $B$ are integers within the range $[0..2,000,000,000]$.
- $K$ is an integer within the range $[1..2,000,000,000]$.
- $A \le B$.

```python
def solution(A, B, K):
    return (B // K) - ((A - 1) // K)
```

```python
import time
import unittest


class TestCountDiv(unittest.TestCase):

    def check_case(self, A, B, K, expected):
        start = time.perf_counter()
        result = solution(A, B, K)
        runtime = time.perf_counter() - start

        print()
        print("=" * 70)
        print(f"Input      : A={A}, B={B}, K={K}")
        print(f"Output     : {result}")
        print(f"Expected   : {expected}")
        print(f"Runtime    : {runtime:.8f}s")
        print(f"Complexity : O(1)")
        print(f"Status     : {'PASS' if result == expected else 'FAIL'}")
        print("=" * 70)

        self.assertEqual(result, expected)

    def test_example(self):
        """Standard example from the problem description."""
        self.check_case(6, 11, 2, 3)

    def test_a_equals_b_divisible(self):
        """Single element range where the element is divisible by K."""
        self.check_case(10, 10, 2, 1)

    def test_a_equals_b_not_divisible(self):
        """Single element range where the element is NOT divisible by K."""
        self.check_case(10, 10, 3, 0)

    def test_a_is_zero(self):
        """Range starting at zero."""
        self.check_case(0, 10, 2, 6)


if __name__ == "__main__":
    suite = unittest.TestLoader().loadTestsFromTestCase(TestCountDiv)
    unittest.TextTestRunner(verbosity=2).run(suite)
```

# MinAvgTwoSlice

Find the minimal average of any slice containing at least two elements.

## Problem Statement
A distributed data pipeline logs performance metrics for a sequential stream of $N$ micro-batches. You are given a non-empty array $A$ of $N$ integers representing the processing latency (in milliseconds) recorded for each consecutive micro-batch.

A continuous execution window is defined by a pair of indices $(P, Q)$ such that $0 \le P < Q < N$. The window contains at least two consecutive micro-batches, and its average processing latency is calculated as:

$$\text{Average}(P, Q) = \frac{\sum_{i=P}^{Q} A[i]}{Q - P + 1}$$

Write a function `solution(A)` that returns the starting index $P$ of the continuous execution window that achieves the minimal average processing latency. If multiple windows share the exact same minimal average, return the smallest starting index $P$.

## Example
Input: $A = [4, 2, 2, 5, 1, 5, 8]$  
Output: `1`  
Explanation: 
- Execution window $(1, 2)$ covers latencies $[2, 2]$ with average $\frac{2 + 2}{2} = 2.0$.
- Execution window $(3, 4)$ covers latencies $[5, 1]$ with average $\frac{5 + 1}{2} = 3.0$.
- Execution window $(1, 4)$ covers latencies $[2, 2, 5, 1]$ with average $\frac{2 + 2 + 5 + 1}{4} = 2.5$.

The window starting at index `1` (range $[1..2]$) yields the minimal overall average latency of `2.0`.

## Constraints
- $N$ is an integer within the range $[2..100,000]$.
- Each element of array $A$ is an integer within the range $[-10,000..10,000]$.

```python
import logging

# Configure logging
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s'
)
logger = logging.getLogger(__name__)

def solution(A):
    """
    Find the index of the slice with the minimum average (length 2 or 3).
    """
    try:
        # --- Exception Handling: Validate input ---
        if A is None:
            raise ValueError("Input array cannot be None")
        
        if not isinstance(A, (list, tuple)):
            raise TypeError(f"Expected list/tuple, got {type(A).__name__}")
        
        if len(A) < 2:
            raise ValueError(f"Array must have at least 2 elements, got {len(A)}")
        
        logger.info(f"Starting solution with array of length {len(A)}")

        min_avg = float('inf')
        min_index = 0
        N = len(A)
        for i in range(N-1):
            # Step 1: calculate length-2 slice average starting at index i
            avg_length_2 = ((A[i] + A[i+1]) / 2)
            # Step 2: update min_avg and min_index if this average is smaller
            if avg_length_2 < min_avg:
                min_avg = avg_length_2
                min_index = i
            # Step 3: if i < N - 2, check lenth-3 slice average starting at index i
            if i < N-2:
                avg_length_3 = ((A[i] + A[i+1] + A[i+2]) / 3)
            # Step 4:: update min_avg and mind_index if this average is smaller
                if avg_length_3 < min_avg:
                    min_avg = avg_length_3
                    min_index = i
    except ValueError as e:
        logger.error(f"Validation error: {e}")
        raise
    except TypeError as e:
        logger.error(f"Type error: {e}")
        raise
    except Exception as e:
        logger.critical(f"Unexpected error: {e}", exc_info=True)
        raise
    return min_index
```

```python
import time
import unittest


class TestMinAvgSlice(unittest.TestCase):

    def check_case(self, A, expected):
        start = time.perf_counter()
        result = solution(A)
        runtime = time.perf_counter() - start

        print()
        print("=" * 70)
        print(f"Input      : {A}")
        print(f"Output     : {result}")
        print(f"Expected   : {expected}")
        print(f"Runtime    : {runtime:.8f}s")
        print(f"Status     : {'PASS' if result == expected else 'FAIL'}")
        print("=" * 70)

        self.assertEqual(result, expected)

    def test_example(self):
        """Standard example - minimum average is at index 1."""
        self.check_case([4, 2, 2, 5, 1, 5, 8], 1)

    def test_min_at_start(self):
        """Minimum average at the beginning."""
        self.check_case([1, 1, 5, 5, 5, 5], 0)

    def test_min_at_end(self):
        """Minimum average at the end."""
        self.check_case([5, 5, 5, 1, 1], 3)

    def test_all_same_values(self):
        """All identical values - any index is valid."""
        result = solution([3, 3, 3, 3])
        self.assertIn(result, [0, 1, 2])

    def test_length_2_min(self):
        """Minimum average comes from a length-2 slice."""
        self.check_case([10, 1, 1, 1], 1)

    def test_length_3_min(self):
        """Minimum average comes from a length-3 slice."""
        self.check_case([1, 2, 3, 10], 0)

    def test_two_elements(self):
        """Array with exactly 2 elements."""
        self.check_case([5, 10], 0)

    def test_three_elements(self):
        """Array with exactly 3 elements."""
        self.check_case([5, 10, 3], 0)

    def test_negative_numbers(self):
        """Array with negative numbers."""
        self.check_case([-10, -5, -2, -1], 0)

    def test_mixed_numbers(self):
        """Mix of positive and negative numbers."""
        self.check_case([-5, 10, -3, 2], 2)


if __name__ == "__main__":
    suite = unittest.TestLoader().loadTestsFromTestCase(TestMinAvgSlice)
    unittest.TextTestRunner(verbosity=2).run(suite)
```
