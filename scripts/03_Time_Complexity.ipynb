# FrogJmp

Count minimal number of jumps from position X to Y.


A frog at position X wants to reach position Y by jumping a fixed distance D.
Find the minimum number of jumps needed to reach or pass Y.

**Example:** `X = 10, Y = 85, D = 30` → answer is `3`
- Jump 1: 10 + 30 = 40
- Jump 2: 40 + 30 = 70
- Jump 3: 70 + 30 = 100 ✅ (≥ 85)

### Constraints
- `1 <= X, Y, D <= 1,000,000,000`
- `X <= Y`

```python
import unittest
import math
import time

def solution(X, Y, D):
    return math.ceil((Y - X) / D)
```

```python
class TestFrogJump(unittest.TestCase):

    def check_case(self, X, Y, D, expected):
        start = time.perf_counter()

        result = solution(X, Y, D)

        end = time.perf_counter()
        runtime = end - start

        print(
            f"Input: X={X}, Y={Y}, D={D} | "
            f"Output: {result} | "
            f"Expected: {expected} | "
            f"Runtime: {runtime:.10f}s | "
            f"Complexity: O(1)"
        )

        self.assertEqual(result, expected)

    def test_example(self):
        self.check_case(10, 85, 30, 3)

    def test_already_at_target(self):
        self.check_case(10, 10, 30, 0)

    def test_exact_jump(self):
        self.check_case(10, 40, 30, 1)

    def test_not_exact_jump(self):
        self.check_case(10, 41, 30, 2)

    def test_large_values(self):
        self.check_case(1, 1_000_000_000, 1, 999_999_999)

    def test_large_d(self):
        self.check_case(1, 1_000_000_000, 1_000_000_000, 1)


if __name__ == "__main__":
    unittest.main(argv=[""], exit=False)
```

# PermMissingElem

Find the missing element in a given permutation.

Given an array A of N distinct integers in the range [1..(N+1)],
exactly one number from that range is missing. Find it.

**Example:** `A = [2, 3, 1, 5]` → answer is `4`

### Constraints
- `0 <= N <= 100,000`
- All elements are distinct
- Each element is in range `[1..(N+1)]`
- Must run efficiently

```python
def solution(A):
    N = len(A)
    complete = set(range(1, N + 2))
    return (complete - set(A)).pop()
```

```python
import unittest
import time

class TestPermMissingElem(unittest.TestCase):

    def check_case(self, A, expected):
        start = time.perf_counter()

        result = solution(A)

        end = time.perf_counter()
        runtime = end - start

        print(
            f"A={A if len(A) <= 10 else 'large array'} | "
            f"Output={result} | "
            f"Expected={expected} | "
            f"Runtime={runtime:.10f}s | "
            f"Time Complexity=O(N) | "
            f"Space Complexity=O(N)"
        )

        self.assertEqual(result, expected)

    def test_example(self):
        self.check_case([2, 3, 1, 5], 4)

    def test_empty_array(self):
        self.check_case([], 1)

    def test_missing_first_element(self):
        self.check_case([2, 3, 4, 5], 1)

    def test_missing_last_element(self):
        self.check_case([1, 2, 3, 4], 5)

    def test_single_element_missing_1(self):
        self.check_case([2], 1)

    def test_single_element_missing_2(self):
        self.check_case([1], 2)

    def test_middle_missing_element(self):
        self.check_case([1, 2, 4, 5, 6], 3)

    def test_large_array(self):
        A = list(range(1, 100_002))
        A.remove(54_321)
        self.check_case(A, 54_321)

suite = unittest.TestLoader().loadTestsFromTestCase(TestPermMissingElem)
unittest.TextTestRunner().run(suite)
```

## Tape Equilibrium

Given an array A of N integers, a split at position P divides the array
into two non-empty parts:
- Left part:  A[0], A[1], ..., A[P-1]
- Right part: A[P], A[P+1], ..., A[N-1]

Find the split position P that minimizes the absolute difference
between the sum of the left part and the sum of the right part.

**Example:** `A = [3, 1, 2, 4, 3]`
- P=1: |3 − 10| = 7
- P=2: |4 − 9|  = 5
- P=3: |6 − 7|  = 1  ← minimum
- P=4: |10 − 3| = 7

Answer: `1`

### Constraints
- `2 <= N <= 100,000`
- `-1,000 <= A[i] <= 1,000`
- P must satisfy `0 < P < N` (both parts must be non-empty)

```python
def solution(A):
    total = sum(A)
    left_sum = A[0]
    right_sum = total - left_sum 
    min_diff = float('inf')

    for i in range(1, len(A)):
        diff = abs(left_sum - right_sum)
        min_diff = min(min_diff, diff)
        left_sum = left_sum + A[i]
        right_sum = total - left_sum

    return min_diff
        

    
```

```python
import unittest
import time

class TestTapeEquilibrium(unittest.TestCase):

    def check_case(self, A, expected):
        start = time.perf_counter()

        result = solution(A)

        end = time.perf_counter()
        runtime = end - start

        print(
            f"A={A if len(A) <= 10 else 'large array'} | "
            f"Output={result} | "
            f"Expected={expected} | "
            f"Runtime={runtime:.10f}s | "
            f"Time Complexity=O(N) | "
            f"Space Complexity=O(1)"
        )

        self.assertEqual(result, expected)

    def test_example(self):
        self.check_case([3, 1, 2, 4, 3], 1)

    def test_two_elements(self):
        self.check_case([10, 20], 10)

    def test_two_equal_elements(self):
        self.check_case([5, 5], 0)

    def test_small_array(self):
        self.check_case([1, 2, 3], 0)

    def test_all_positive(self):
        self.check_case([5, 6, 2, 4, 1], 4)

    def test_all_negative(self):
        self.check_case([-3, -1, -2, -4, -3], 1)

    def test_mixed_positive_negative(self):
        self.check_case([3, -1, 2, -4, 3], 1)

    def test_large_array(self):
        A = [1] * 100_000
        self.check_case(A, 0)


suite = unittest.TestLoader().loadTestsFromTestCase(TestTapeEquilibrium)
unittest.TextTestRunner().run(suite)
```
