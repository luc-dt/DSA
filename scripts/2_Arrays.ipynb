## Cyclic Rotation

Given an array A of N integers, a **rotation** shifts every element one position
to the right, and moves the last element to the front.

For example:
- `[3, 8, 9, 7, 6]` rotated once → `[6, 3, 8, 9, 7]`

### Task
Given array A and integer K, return the array after K rotations.

**Examples:**
- `A = [3, 8, 9, 7, 6], K = 3` → `[9, 7, 6, 3, 8]`
- `A = [1, 2, 3, 4], K = 4` → `[1, 2, 3, 4]` (full cycle, back to original)
- `A = [0, 0, 0], K = 1` → `[0, 0, 0]`

### Constraints
- `0 <= N, K <= 100`
- `-1000 <= A[i] <= 1000`

```python
def solution(A, K):
  if len(A) == 0:
    return 0
  # Optimization: handle cases where K is larger than the array length
  K = K % len(A)

  return A[-K:] + A[:-K]
```

```python
A = [1, 2, 3]
K = 3
solution(A, K)
```

```python
A = [3, 8, 9, 7, 6]
K = 3
solution(A, K)
```

## OddOccurencesinArray

Given a non-empty array A of N integers where N is always odd, every element
appears an even number of times — except for exactly one unpaired element.

For example:
- `A = [9, 3, 9, 3, 9, 7, 9]`
  - `9` appears 4 times (paired)
  - `3` appears 2 times (paired)
  - `7` appears 1 time → **unpaired**

### Task
Return the value of the unpaired element.

**Example:** `A = [9, 3, 9, 3, 9, 7, 9]` → answer is `7`

### Constraints
- `1 <= N <= 1,000,000` (N is always odd)
- `1 <= A[i] <= 1,000,000,000`
- All elements except one appear an even number of times

```python
def solution(A):
    # Initialize an empty set to track elements without a pair
    unpaired_element = set()

    # Iterate through each number in the input array
    for num in A:
        if num not in unpaired_element:
          unpaired_element.add(num)
          print('add to set: ', num, 'current_set: ', unpaired_element)
        else:
          unpaired_element.remove(num)
          print('remove from set: ', num, 'current_set: ', unpaired_element)

    # The only element left in the set is the unpaired one
    return list(unpaired_element)[0]
```

```python
A = [9, 3, 9, 3, 9, 7, 9]
solution(A)
```

## Maximum Subarray Sum

"Contiguous" means elements must be consecutive in the original array —
no skipping or reordering allowed.

### Task
Given an array of integers, find the contiguous subarray that produces
the largest sum and return that sum.

**Examples:**
- `[-2, 1, -3, 4, -1, 2, 1, -5, 4]` → answer is `6` (subarray `[4, -1, 2, 1]`)
- `[-3, -1, -2]` → answer is `-1` (least negative element)
- `[5]` → answer is `5`

### Constraints
- `1 <= n <= 100,000`
- `-10,000 <= arr[i] <= 10,000`
- If all elements are negative, return the maximum (least negative) element
- Must run in **O(n)** time


```python
def max_subarray_sum(data_dict):
    """ 
    :type nums: List[int] 
    :rtype: int 
    """
    nums = data_dict["nums"]
    
    current_sum = nums[0]
    max_sum = nums[0]
    
    for num in nums[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
    if current_sum < 0:
        return -1
    return max_sum
    
```

## Common Elements with Repeated Values

Easy

### Task
Find common elements in 2 given lists and return all aspects even if they are repeated.

### Constraints
- The input variables list1 and list2 should be of type List[int].
- The input lists can have any length.
- The elements in the input lists can be repeated.
- The elements in the input lists can be in any order.
- The input lists can contain both positive and negative integers.
- The input lists can contain zero as an element.
- The input lists can be empty.


```python
def common_elements(input):
    list1 = input["list1"]
    list2 = input["list2"]

    counter1 = {}
    counter2 = {}
    
    for item in list1:
        if item in counter1:
            counter1[item] += 1
        else:
            counter1[item] = 1
    
    for item in list2:
        if item in counter2:
            counter2[item] += 1
        else:
            counter2[item] = 1
            
    common = {}
    for key in counter1:
        if key in counter2:
            common[key] = min(counter1[key], counter2[key])
    
    result = []
    for key, count in common.items():
        result.extend([key] * count)
        
    return result
```

```python
# Solution 2

from collections import Counter 
from typing import List 

def find_common_elements(list1: List[int], list2: List[int]) -> List[int]:
    """
    Finds common elements between two lists, preserving repeated elements
    based on the minimum frequency in both lists.
    """
    # 1. Edge case: quick return if either list is empty 
    if not list1 or not list2:
        return []

    # 2. Count frequencies of elements in both lists
    count1 = Counter(list1)
    count2 = Counter(list2)

    # 3. Intersect counters (take min frequency of common elements)
    intersection = count1 & count2

    return list(intersection.elements())


# --- Quick Verification ---
if __name__ == "__main__":
    print(find_common_elements([1, 2, 2, 1], [2, 2]))       # Output: [2, 2]
    print(find_common_elements([4, 9, 5, 4], [9, 4, 9, 8, 4])) # Output: [4, 4, 9]
    print(find_common_elements([-1, 0, 0], [0, 0, 5]))     # Output: [0, 0]

```

## Finding Target Sums

Medium

### Task
You are given a list of numbers and a target sum. The objective is to find all consecutive subarrays whose total matches the target, then sort these subarrays by ascending length and select only up to a maximum number from that sorted set. This guarantees that the smallest subarrays appear first and that no more than the desired limit of results are included.

### Constraints
- The input list nums can have any length (including empty).
- The elements in nums can be positive, negative, or zero.
- The target sum target_sum can be any integer (positive, negative, or zero).
- The value of n can be any non-negative integer (including zero).
- Each interval in the output is a consecutive subsequence of nums.
- The final output list can have at most n intervals.


```python
from unittest import main
def find_intervals(input):
    """
    :type input: Dict[str, Any]
    :rtype: List[List[int]]
    """

    nums = input["nums"]
    target_sum = input["target_sum"]
    n = input["n"]

    tmp = []

    for i in range(len(nums)):
        total = 0
        for j in range(i, len(nums)):
            total += nums[j]
            if total == target_sum:
                tmp.append(nums[i:j+1])

    return sorted(tmp, key=len)[:n]
```

```python
import unittest
class TestFindIntervals(unittest.TestCase):

    def test_standard_positive_numbers(self):
        """Case 1: Standard positive sequence"""
        payload = {"n": 2, "nums": [1, 2, 3, 4, 5], "target_sum": 5}
        expected = [[5], [2, 3]]
        self.assertEqual(find_intervals(payload), expected)

    def test_zero_sum_with_negatives(self):
        """Case 2: Target sum is 0 with alternating signs"""
        payload = {"n": 3, "nums": [1, -1, 2, -2, 3, -3, 4], "target_sum": 0}
        # Shortest arrays that sum to 0
        expected = [[1, -1], [2, -2], [3, -3]]
        self.assertEqual(find_intervals(payload), expected)

    def test_varying_lengths_and_duplicates(self):
        """Case 3: Multiple matching subsets of varying lengths"""
        payload = {"n": 2, "nums": [10, 20, 10, -10, 5, 5, 5, 5, 10], "target_sum": 15}
        # [10, -10, 5, 5, 5] len 5, [5, 5, 5] len 3, [5, 10] len 2
        # Shortest 2 will be [5, 10] and [5, 5, 5]
        expected = [[5, 10], [5, 5, 5]]
        self.assertEqual(find_intervals(payload), expected)

    def test_no_match(self):
        """Edge Case: No subarrays sum up to the target"""
        payload = {"n": 2, "nums": [1, 2, 3], "target_sum": 100}
        expected = []
        self.assertEqual(find_intervals(payload), expected)


if __name__ == "__main__":
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

## Digit Summing Challenge

Easy

### Task
Sum the digits of every number in a list.

### Constraints
- The input variable numbers is a list of integers.
- The integers in the list can be positive or negative.
- The integers in the list can have multiple digits.
- The length of the list can be any positive integer.
- The integers in the list can be within the range of the integer data type.


```python
def sum_digits(numbers):
    """ 
    :type numbers: List[int]
    :rtype: List[int]
    """
    return [sum(int(digit) for digit in str(abs(num))) for num in numbers]
```

```python
import unittest
class TestSumDigits(unittest.TestCase):

    def test_positive_and_negative_numbers(self):
        numbers = [12345, 67890, -12345, -67890]
        expected = [15, 30, 15, 30]
        self.assertEqual(sum_digits(numbers), expected)

    def test_zero_and_repeated_digits(self):
        numbers = [0, -111, 2222, -33333, 444444]
        expected = [0, 3, 8, 15, 24]
        self.assertEqual(sum_digits(numbers), expected)

    def test_mixed_large_numbers(self):
        numbers = [-1001, 505, -202, 99999, 123456789]
        expected = [2, 10, 4, 45, 45]
        self.assertEqual(sum_digits(numbers), expected)

    def test_empty_list(self):
        numbers = []
        expected = []
        self.assertEqual(sum_digits(numbers), expected)


if __name__ == "__main__":
    unittest.main(argv=[""], exit=False)
```

## Unique Zero Sum Triplets in Array

Medium

### Task
Given an array of integers A, find all triplets (i, j, k) in the array such that i != j, i != k, j != k, and i + j + k = 0. This means that while the indices of the triplet must be distinct, the values at these indices are allowed to be the same if they occur at different positions. Each triplet in the result must be sorted in non-decreasing order, and the list of triplets should also be sorted in lexicographic order. The solution should not contain duplicate triplets.

### Constraints
- The input variable nums is a list of integers.
- The length of nums can be any non-negative integer.
- The integers in nums can be positive, negative, or zero.
- The integers in nums can be repeated.
- The solution assumes that the input list nums is not empty.


```python
# Problem: Unique Zero Sum Triplets

# * Find all groups of 3 numbers in the array that sum to zero
# * The 3 numbers must come from distinct indices (but values can repeat if at different positions)
# * Each triplet must be sorted in non-decreasing order → [-1, 0, 1] not [1, -1, 0]
# * The final list of triplets must be sorted in lexicographic order → compare element by element left to right
# * No duplicate triplets in the output

def find_triplets(nums):
    """ 
    :type nums: List[int]
    :rtype: List[List[int]]
    """
    # use a set to automatically handle duplicate triplets
    set_triple = set()

    # fix the first element of the triplet
    for i in range(len(nums)):

        # fix the second element, always ahead of i to avoid reuse
        for j in range(i+1, len(nums)):

            # fix the third element, always ahead of j to avoid reuse
            for k in range(j+1, len(nums)):

                # check if the three elements sum to zero
                if (nums[i] + nums[j] + nums[k]) == 0:

                    # sort the triplet to ensure non-decreasing order
                    # convert to tuple because lists are not hashable
                    sorted_triplet = tuple(sorted([nums[i], nums[j], nums[k]]))

                    # add to set — duplicates are ignored automatically
                    set_triple.add(sorted_triplet)

    # convert set of tuples back to list of lists
    return list(set_triple)
```

```python
import unittest
class TestFindTriplets(unittest.TestCase):

    def test_case_1(self):
        nums = [1, -1, 0, 2, -2, 3, -3]

        expected = {
            (-1, 0, 1),
            (-2, 0, 2),
            (-3, 0, 3),
            (-2, -1, 3),
            (-3, 1, 2)
        }

        self.assertEqual(set(find_triplets(nums)), expected)

    def test_case_2(self):
        nums = [3, 0, -2, 1, 1, 2, -1, -4, 2]

        expected = {
            (-1, 0, 1),
            (-4, 1, 3),
            (-2, 1, 1),
            (-4, 2, 2),
            (-2, -1, 3),
            (-2, 0, 2)
        }

        self.assertEqual(set(find_triplets(nums)), expected)

    def test_case_3(self):
        nums = [8, -25, 18, 2, -10, 6, -8, 4]

        expected = {
            (-10, 4, 6),
            (-8, 2, 6),
            (-10, -8, 18),
            (-10, 2, 8)
        }

        self.assertEqual(set(find_triplets(nums)), expected)

    def test_no_triplets(self):
        nums = [1, 2, 3, 4]

        expected = set()

        self.assertEqual(set(find_triplets(nums)), expected)

    def test_all_zeroes(self):
        nums = [0, 0, 0, 0]

        expected = {
            (0, 0, 0)
        }

        self.assertEqual(set(find_triplets(nums)), expected)


if __name__ == "__main__":
    unittest.main(argv=[""], exit=False)
```

```python
# solution 2 : with two pointers

def find_triplets_p(nums):
    """
    Find all unique triplets whose sum is 0.
    :type nums: List[int]
    :rtype: List[List[int]]
    """

    # Sort the list first
    # This allows use to use the two-pointer technique. 
    nums.sort()

    # Use a set to store unique triplets. 
    # Tuple is used because list cannot be stored in a set. 
    triplets = set() 

    # Fix the first number nums[i]
    # We stop at len(nums) -2 because we need at least 2 more numbers after i. 
    for i in range(len(nums) - 2):
        # left starts right after i 
        # right starts at the end of the list. 
        left, right = i + 1, len(nums) - 1 

        # Move left and right pointers toward each other. 
        while left < right:
            # calculate the sum of the current triplet 
            total = nums[i] + nums[left] + nums[right]

            # if total is too small, move left to the right
            # because the list is sorted and we need a bigger number. 
            if total < 0:
                left += 1 
            
            # if total is too large, move right to the left 
            # because the list is sorted and we need a smaller number. 
            elif total > 0:
                right -=  1

            # if total is 0, we found a valid triplet. 
            else:
                triplets.add((nums[i], nums[left], nums[right]))

                # move both pointers to continue searching. 
                left += 1 
                right -= 1 

    # Convert each tuple back to list before returning 
    return [list(triplet) for triplet in triplets]
```

```python
nums1 = [1, -1, 0, 2, -2, 3, -3]
nums2 = [3, 0, -2, 1, 1, 2, -1, -4, 2]
nums3 = [8, -25, 18, 2, -10, 6, -8, 4]
```

```python
print(find_triplets_p(nums1))
print(find_triplets_p(nums2))
print(find_triplets_p(nums3))
```

## Minimum Coins for Total Amount

Medium

### Task
Given an amount of money and a list of coin denominations, what is the minimum number of coins you must select to make up that amount, assuming an unlimited number of coins from each denomination?

### Constraints
- The input variable amount is an integer representing the target amount of money.
- The input variable coins is a list of integers representing the available coin denominations.
- The list coins can have any length.
- The elements in the list coins can be positive integers.
- The elements in the list coins can be in any order.
- The target amount amount can be any non-negative integer.
- The target amount amount can be equal to zero.
- The target amount amount can be greater than zero.
- The target amount amount can be greater than the maximum value that can be represented by an integer.


```python
# Version 1: just show the number of coins need to be equally amount
def min_coins(input):
    """
    :type input: Dict[str, Union[int, List[int]]]
    :rtype: int
    """

    amount = input["amount"]
    coins = input["coins"]

    # Build a table where dp[i] = minimum coins needed to make amount i
    # Initialize all values to infinity (meaning "not yet reachable")
    dp = [float("inf")] * (amount + 1)
    
    # Base case: 0 coins needed to make amount 0
    dp[0] = 0

    
    # Build up the solution from amount 1 to target amount
    for i in range(1, amount + 1):
       # Try every coin denomination
       for c in coins:
           # Only use coin c if it doesn't exceed current amount
           if c <= i:
              # If we use coin c, we need 1 + minimum coins for remaining amount (i - c)
              # Keep the best (minimum) option seen so far
              dp[i] = min(dp[i], dp[i - c] + 1)

    # If dp[amount] is still infinity, no combination of coins can make the amount         
    return -1 if dp[amount] == float("inf") else dp[amount]
    
```

```python
# Version 2: return the minimum number of coins and the combination used to reach the target amount
def min_coins_2(input):
    amount = input["amount"]
    coins = input["coins"]

    # Build a table where dp[i] = minimum coins needed to make amount i
    # Initialize all values to infinity (meaning "not yet reachable")
    dp = [float("inf")] * (amount + 1)

    # Track which coin was used to achieve the optimal solution for each amount
    last_coin = [-1] * (amount + 1)

    # Base case: 0 coins needed to make amount 0
    dp[0] = 0

    # Build up the solution from amount 1 to target amount
    for i in range(1, amount + 1):
        # Try every coin denomination
        for c in coins:
            # Only use coin c if it doesn't exceed current amount
            if c <= i:
                # If using coin c gives a better solution, update dp[i]
                if dp[i - c] + 1 < dp[i]:
                    dp[i] = dp[i - c] + 1
                    # Record which coin gave the best result for amount i
                    last_coin[i] = c

    # If dp[amount] is still infinity, no combination of coins can make the amount
    if dp[amount] == float("inf"):
        return {
            "min_number_of_coins": -1,
            "coins_used": []
        }

    # Backtrack from amount to 0 to reconstruct which coins were used
    result = []
    current_amount = amount

    while current_amount > 0:
        # Find which coin was used to reach current_amount
        coin = last_coin[current_amount]
        result.append(coin)
        # Subtract that coin to move to the previous subproblem
        current_amount -= coin

    return {
        "min_number_of_coins": dp[amount],
        "coins_used": result
    }
```

```python
import unittest

class TestMinCoins2(unittest.TestCase):

    def test_min_coins_with_valid_combination(self):
        input = {"coins": [1, 2, 5], "amount": 11}
        result = min_coins_2(input)

        self.assertEqual(result["min_number_of_coins"], 3)
        self.assertEqual(sum(result["coins_used"]), 11)
        self.assertEqual(sorted(result["coins_used"]), [1, 5, 5])

    def test_min_coins_impossible(self):
        input = {"coins": [2], "amount": 3}
        result = min_coins_2(input)

        self.assertEqual(result["min_number_of_coins"], -1)
        self.assertEqual(result["coins_used"], [])

    def test_amount_zero(self):
        input = {"coins": [1, 2, 3], "amount": 0}
        result = min_coins_2(input)

        self.assertEqual(result["min_number_of_coins"], 0)
        self.assertEqual(result["coins_used"], [])


unittest.main(argv=[""], exit=False)
```
