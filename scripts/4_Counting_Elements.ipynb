# FrogRiverOne
Find the earliest time when a frog can jump to the other side of a river.


A frog wants to cross a river from position `0` to the opposite bank at position `X + 1`. The frog can only cross if leaves cover every single integer position from `1` to `X`.

You are given an integer `X` and an array `A` of `N` integers. `A[K]` represents the position where a leaf falls at time `K` (measured in seconds). Leaves do not move once they land.

Find the **earliest time** (index `K`) when all positions from `1` to `X` are covered by leaves. If it is impossible for the frog to cross, return `-1`.

### Example

`X = 5`, `A = [1, 3, 1, 4, 2, 3, 5, 4]` → answer is `6`

* Time 0 to 5: Positions 1, 2, 3, and 4 are covered, but position 5 is empty.
* Time 6: A leaf falls at position 5. Every position from 1 to 5 now has a leaf. ✅

### Constraints

* `1 <= N, X <= 100,000`
* Each element of array `A` is an integer within the range `[1..X]`

````python
import unittest
import time
def solution(X, A):
    covered_positions = set()
    for time, position in enumerate(A):
        covered_positions.add(position)
        if len(covered_positions) == X:
            return time
    return -1

````

````python
class TestFrogRiverOne(unittest.TestCase):

    def check_case(self, X, A, expected):
        start = time.perf_counter()

        result = solution(X, A)

        end = time.perf_counter()
        runtime = end - start

        # Preview array for clean log output
        a_preview = str(A) if len(A) <= 10 else f"{A[:10]}... [Total elements: {len(A)}]"

        print(
            f"Input: X={X}, A={a_preview} | "
            f"Output: {result} | "
            f"Expected: {expected} | "
            f"Runtime: {runtime:.10f}s | "
            f"Complexity: O(N)"
        )

        self.assertEqual(result, expected)

    def test_example(self):
        """Standard problem example case."""
        self.check_case(5, [1, 3, 1, 4, 2, 3, 5, 4], 6)

    def test_impossible_missing_leaves(self):
        """Leaves fall continuously, but a critical position (3) is never covered."""
        self.check_case(3, [1, 2, 1, 2, 1, 2], -1)

    def test_minimal_case_success(self):
        """Smallest possible valid configuration."""
        self.check_case(1, [1], 0)

    def test_minimal_case_fail(self):
        """Array runs out of events before reaching target X."""
        self.check_case(2, [1], -1)

    def test_complete_at_the_very_end(self):
        """The last missing position lands on the final element of the array."""
        self.check_case(3, [1, 2, 2, 1, 3], 4)

    def test_large_scale_performance(self):
        """Validates O(N) constraint scaling up to 100,000 items."""
        X = 100_000
        A = list(range(1, X + 1))  # 1, 2, 3 ... 100,000
        self.check_case(X, A, X - 1)


if __name__ == "__main__":
    unittest.main(argv=[""], exit=False)
````

# PermCheck

Check whether array A is a permuatation

An array `A` containing `N` integers is considered a valid permutation if it contains every integer from `1` to `N` exactly once, without any omissions or duplicates.

Write a function:

```python
def solution(A):

```

that returns `1` if array `A` is a valid permutation, and `0` if it is not.

### Examples

* **Example 1:** `A = [4, 1, 3, 2]` → answer is `1`
* The array contains `N = 4` elements. Every number from 1 to 4 appears exactly once. ✅


* **Example 2:** `A = [4, 1, 3]` → answer is `0`
* The array contains `N = 3` elements, but the number 2 is missing, and 4 is out of bounds for a permutation of size 3. ✗



### Constraints

* `1 <= N <= 100,000`
* Each element in array `A` is an integer within the range `[1..1,000,000,000]`

````python
def solution(A):
    perfect_sequence = set(range(1, len(A) + 1))
    input_set = set(A)

    if perfect_sequence == input_set:
        return 1
    return 0
````

````python
import time
import unittest
class TestPermCheck(unittest.TestCase):

    def check_case(self, A, expected):
        start = time.perf_counter()
        result = solution(A)
        end = time.perf_counter()
        runtime = end - start

        a_preview = (
            str(A) if len(A) <= 10 else f"{A[:10]}... [Total elements: {len(A)}]"
        )

        print(
            f"Input: A={a_preview} | "
            f"Output: {result} | "
            f"Expected: {expected} | "
            f"Runtime: {runtime:.10f}s | "
            f"Complexity: O(N)"
        )

        self.assertEqual(result, expected)

    def test_valid_permutation(self):
        """Standard valid permutation."""
        self.check_case([4, 1, 3, 2], 1)

    def test_invalid_missing_element(self):
        """Elements are missing, leaving a gap."""
        self.check_case([4, 1, 3], 0)

    def test_single_element_valid(self):
        """Smallest possible valid permutation."""
        self.check_case([1], 1)

    def test_single_element_invalid(self):
        """Single element out of range."""
        self.check_case([2], 0)


if __name__ == "__main__":
    # This targets ONLY the TestPermCheck class
    suite = unittest.TestLoader().loadTestsFromTestCase(TestPermCheck)
    unittest.TextTestRunner().run(suite)
````

# MaxCounters

Calculate the values of counters after applying all alternating operations: increase counter by 1; set value of all counters to current maximum.

You are tracking `N` independent metrics, all initially set to `0`. You are given an event stream represented by an array `A` of `M` integers. Each integer triggers one of two operations:

* **Increment:** If `1 <= A[K] <= N`, increment the metric at position `A[K]` (1-indexed) by 1.
* **Global Sync:** If `A[K] == N + 1`, update every single metric to equal the maximum value achieved by any metric so far.

Write a function:

```python
def solution(N, A):

```

that returns a list of integers representing the final values of all `N` metrics after processing the entire event stream.

### Example

`N = 5`, `A = [3, 4, 4, 6, 1, 4, 4]` → answer is `[3, 2, 2, 4, 2]`

* Operations 0, 1, 2: Metrics become `[0, 0, 1, 2, 0]`. Max value is 2.
* Operation 3 (`A[3] = 6`): Global Sync sets all metrics to 2 → `[2, 2, 2, 2, 2]`.
* Operations 4, 5, 6: Final state is `[3, 2, 2, 4, 2]`. ✅

### Constraints

* `1 <= N, M <= 100,000`
* Each element in array `A` is an integer within the range `[1..N + 1]`

---


````python
def solution(N, A):
    counters = [0] * N

    for step, op in enumerate(A):
        if op <= N:
            counters[op - 1] += 1
            action = f"increase({op})"
        else:
            counters = [max(counters)] * N
            action = "max counter"
        print(f"Step {step}: {action:12} -> {tuple(counters)}")

    return counters
````

````python
A =  [3, 4, 4, 6, 1, 4, 4]
N = 5
solution(N, A)
````

````python
import time
import unittest


class TestMaxCounters(unittest.TestCase):

    def setUp(self):
        """Print a separator before each test."""
        print("\n" + "=" * 80)

    def tearDown(self):
        """Print a separator after each test."""
        print("=" * 80)

    def check_case(self, N, A, expected):
        """Run a test case, measure runtime, and verify the result."""

        start = time.perf_counter()
        result = solution(N, A)
        end = time.perf_counter()
        runtime = end - start

        a_preview = (
            str(A)
            if len(A) <= 10
            else f"{A[:10]}... [Total operations: {len(A)}]"
        )

        print(
            f"N={N} | "
            f"A={a_preview} | "
            f"Output={result} | "
            f"Expected={expected} | "
            f"Runtime={runtime:.10f}s | "
            f"Complexity: O(N × M) (Naive)"
        )

        self.assertEqual(result, expected)

    def test_codility_example(self):
        """Official example from Codility."""
        self.check_case(
            5,
            [3, 4, 4, 6, 1, 4, 4],
            [3, 2, 2, 4, 2]
        )

    def test_only_increase_operations(self):
        """Only increase operations."""
        self.check_case(
            3,
            [1, 2, 3, 1, 2, 3],
            [2, 2, 2]
        )

    def test_only_max_counter_operations(self):
        """Only max counter operations."""
        self.check_case(
            4,
            [5, 5, 5],
            [0, 0, 0, 0]
        )

    def test_single_counter(self):
        """Single counter."""
        self.check_case(
            1,
            [1, 2, 1, 2, 1],
            [3]
        )

    def test_multiple_max_counter(self):
        """Multiple max counter operations."""
        self.check_case(
            2,
            [1, 3, 2, 3, 1],
            [3, 2]
        )

    def test_max_counter_at_end(self):
        """Max counter is the final operation."""
        self.check_case(
            3,
            [1, 2, 2, 4],
            [2, 2, 2]
        )


if __name__ == "__main__":
    suite = unittest.TestLoader().loadTestsFromTestCase(TestMaxCounters)
    unittest.TextTestRunner(verbosity=2).run(suite)
````

# MissingInteger

Find the smallest positive integer that does not occur in a given sequence.

You are managing a data pipeline where new ingestion tasks must be assigned the lowest available positive integer ID (starting from 1). You are given a snapshot of currently occupied IDs represented by an array `A` of `N` integers. Because of legacy system noise, the array may contain zero, duplicate, or negative values, which do not represent valid active IDs.

Write a function:

```python
def solution(A):

```

that returns the smallest positive integer ID (greater than 0) that is currently open (does not appear in `A`).

### Example

* `A = [1, 3, 6, 4, 1, 2]` → answer is `5` (IDs 1, 2, 3, 4, and 6 are occupied; 5 is the lowest free ID).
* `A = [1, 2, 3]` → answer is `4` (All sequential IDs up to 3 are taken).
* `A = [-1, -3]` → answer is `1` (No valid positive IDs are occupied, so 1 is the default start).

### Constraints

* `N` is an integer within the range `[1..100,000]`
* Each element in array `A` is an integer within the range `[-1,000,000..1,000,000]`

````python
def solution(A):
    S = set(A)
    candidate = 1
    while candidate in S:
        candidate += 1
    return candidate
````

````python
import time
import unittest


class TestMissingInteger(unittest.TestCase):

    def setUp(self):
        """Print a separator before each test."""
        print("\n" + "=" * 80)

    def tearDown(self):
        """Print a separator after each test."""
        print("=" * 80)

    def check_case(self, A, expected):
        """Run a test case, measure runtime, and verify the result."""

        start = time.perf_counter()
        result = solution(A)
        end = time.perf_counter()

        runtime = end - start

        a_preview = (
            str(A)
            if len(A) <= 15
            else f"{A[:15]}... [Total elements: {len(A)}]"
        )

        print(
            f"Input={a_preview} | "
            f"Output={result} | "
            f"Expected={expected} | "
            f"Runtime={runtime:.10f}s | "
            f"Complexity: O(N)"
        )

        self.assertEqual(result, expected)

    # ==========================================================
    # Official Codility Examples
    # ==========================================================

    def test_codility_example_1(self):
        """Official example: missing number in the middle."""
        self.check_case(
            [1, 3, 6, 4, 1, 2],
            5
        )

    def test_codility_example_2(self):
        """Official example: consecutive positives."""
        self.check_case(
            [1, 2, 3],
            4
        )

    def test_codility_example_3(self):
        """Official example: all negatives."""
        self.check_case(
            [-1, -3],
            1
        )

    # ==========================================================
    # Edge Cases
    # ==========================================================

    def test_single_positive(self):
        """Single positive number."""
        self.check_case(
            [1],
            2
        )

    def test_single_negative(self):
        """Single negative number."""
        self.check_case(
            [-1],
            1
        )

    def test_zero_only(self):
        """Zero is not a positive integer."""
        self.check_case(
            [0],
            1
        )

    def test_duplicates(self):
        """Duplicate values should not affect the answer."""
        self.check_case(
            [1, 1, 2, 2, 3, 3],
            4
        )

    def test_missing_one(self):
        """Smallest positive integer is missing."""
        self.check_case(
            [2, 3, 4],
            1
        )

    def test_large_gap(self):
        """Large gap after the first value."""
        self.check_case(
            [1, 1000],
            2
        )

    def test_mixed_numbers(self):
        """Mix of negative, zero and positive values."""
        self.check_case(
            [-5, -1, 0, 2, 3],
            1
        )

    def test_unsorted(self):
        """Input is completely unsorted."""
        self.check_case(
            [4, 2, 1, 6, 5],
            3
        )

    def test_all_duplicates(self):
        """All elements are identical."""
        self.check_case(
            [2, 2, 2, 2],
            1
        )

    def test_missing_after_many_duplicates(self):
        """Many duplicates with one missing value."""
        self.check_case(
            [1, 2, 2, 3, 3, 4, 4],
            5
        )


if __name__ == "__main__":
    suite = unittest.TestLoader().loadTestsFromTestCase(TestMissingInteger)
    unittest.TextTestRunner(verbosity=2).run(suite)
````
